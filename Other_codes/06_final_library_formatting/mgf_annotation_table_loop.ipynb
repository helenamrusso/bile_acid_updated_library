```python
from pathlib import Path
import pandas as pd

# Build one annotation table per MGF and save it next to the input file.
for mgf_label, input_file in mgf_pre_falcon.items():
    scans = []
    origin = []
    massql_pepmass = []
    pepmass = []
    original_path = []
    original_scan = []
    new_filename = []
    new_scan = []
    massql_intensity = []

    with open(input_file, "rb") as fin:
        input_lines = fin.readlines()

    for line in input_lines:
        decoded_line = line.decode("utf-8", errors="ignore")
        if decoded_line.startswith("SCANS="):
            scans.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("PEPMASS="):
            pepmass.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_ORIGIN="):
            origin.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_PEPMASS="):
            massql_pepmass.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_ORIGINAL_PATH="):
            original_path.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_SCAN="):
            original_scan.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_NEW_FILENAME="):
            new_filename.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_NEW_SCAN="):
            new_scan.append(decoded_line.strip().split("=", 1)[1])
        elif decoded_line.startswith("MASSQL_I="):
            massql_intensity.append(decoded_line.strip().split("=", 1)[1])

    # Keep only rows present across all required columns to avoid length mismatch.
    n_rows = min(
        len(scans),
        len(origin),
        len(pepmass),
        len(massql_pepmass),
        len(original_path),
        len(original_scan),
        len(new_filename),
        len(new_scan),
        len(massql_intensity),
    )

    if n_rows == 0:
        print(f"[{mgf_label}] No annotation fields found in: {input_file}")
        continue

    df = pd.DataFrame({
        "scan_combined_mgf": scans[:n_rows],
        "origin_repository": origin[:n_rows],
        "pepmass": pepmass[:n_rows],
        "massql_precmz": massql_pepmass[:n_rows],
        "original_path": original_path[:n_rows],
        "original_scan": original_scan[:n_rows],
        "new_filename": new_filename[:n_rows],
        "new_scan": new_scan[:n_rows],
        "massql_intensity": massql_intensity[:n_rows],
    })

    input_path = Path(input_file)
    output_file = input_path.with_name(f"{input_path.stem}_annotation_table.tsv")
    df.to_csv(output_file, sep="\t", index=False)
    print(f"[{mgf_label}] Wrote {len(df)} rows -> {output_file}")
```
